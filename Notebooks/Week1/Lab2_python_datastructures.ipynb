### List comprehensions

```python
salaries = [45000, 62000, 78000, 51000, 95000, 43000, 88000, 71000]
raised_salaries = [s * 1.10 for s in salaries]
print(f"Orginal Salaries: {salaries}")
print(f"raised Salaries: {raised_salaries}")
high_earners = [s for s in salaries if s > 60000]
print(high_earners)
```

### Dictionaries and dict comprehensions

```python
employees = {
  'E001': {'name': 'Sara Ahmed', 'dept': 'Data Science', 'salary': 78000},
  'E002': {'name': 'Bilal Khan', 'dept': 'Engineering', 'salary': 92000},
  'E003': {'name': 'Aisha Malik', 'dept': 'Data Science', 'salary': 85000},
}
names = {emp_id: info['name'] for emp_id, info in employees.items()}
print(names)
ds_employees = {k: v for k, v in employees.items() if v['dept'] == 'Data Science'}
print(ds_employees)
```

### Lambda functions

```python
classify_salary = lambda s: 'High' if s > 80000 else ('Mid' if s > 55000 else 'Low')
salary_labels = list(map(classify_salary, salaries))
sorted_employees = sorted(employees.items(), key=lambda x: x[1]['salary'], reverse=True)
print(sorted_employees)
```

### Performance: NumPy vs Python list

```python
import numpy as np, time
size = 10_000_000
py_list = list(range(size))
start = time.time()
result = [x * 2 for x in py_list]
print(f'Python list: {time.time() - start:.3f} seconds')

np_array = np.arange(size)
start = time.time()
result = np_array * 2
print(f'NumPy array: {time.time() - start:.3f} seconds')
```

### NumPy statistics and boolean indexing

```python
np.random.seed(42)
purchases = np.random.normal(loc=120, scale=45, size=1000)
purchases = np.clip(purchases, 5, 500)

print(f'Mean: ${purchases.mean():.2f}')
print(f'Median: ${np.median(purchases):.2f}')
high_value = purchases[purchases > 200]
print(f'Over $200: {len(high_value)} ({len(high_value)/len(purchases)*100:.1f}%)')
discounted = purchases * 0.85
```

### Matrix operations

```python
matrix = np.array([[2, 0, 1], [3, 1, 4], [5, 2, 0]])
print('Transpose:', matrix.T)
identity = np.eye(3)
result = np.dot(matrix, identity)
print('Matrix @ Identity = Matrix:', np.allclose(result, matrix))
```
