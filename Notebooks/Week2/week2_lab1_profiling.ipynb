### Step 1 — Load & preview

```python
df = sns.load_dataset('titanic')
print(df.head())
print(df.shape)
```

### Step 2 — Shape & types

```python
print('Shape:', df.shape)
print(df.info())
print('Memory (KB):', df.memory_usage(deep=True).sum() / 1024)
```

###  Step-3 Missing values & duplicates

```python
missing = df.isnull().sum().sort_values(ascending=False)
missing_pct = (missing / len(df) * 100).round(2)
print(pd.DataFrame({'Missing Count': missing, 'Missing %': missing_pct}))
print('Duplicate rows:', df.duplicated().sum())
```

### Step 4 — Descriptive statistics
Stats + skewness + kurtosis

```python
print(df.describe().T)
numeric_cols = df.select_dtypes(include='number').columns
print('Skewness:')
print(df[numeric_cols].skew().round(3))
print('Kurtosis:')
print(df[numeric_cols].kurtosis().round(3))
```

### Step 5 - Categorical summary
Value counts per category

```python
for col in ['survived', 'sex', 'pclass', 'embarked']:
    print(f'\n--- {col} ---')
    print(df[col].value_counts())
    print(f'Unique values: {df[col].nunique()}')
```
